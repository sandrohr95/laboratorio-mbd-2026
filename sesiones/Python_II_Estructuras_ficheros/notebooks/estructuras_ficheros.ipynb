## 1. Listas

Una lista (`list`) es una colección **ordenada** y **modificable** de elementos. Se escribe entre corchetes y puede contener elementos de cualquier tipo, incluso otras listas. Los índices y el *slicing* funcionan igual que en las cadenas.

````python
ciudades = ["Málaga", "Sevilla", "Granada"]

print(ciudades[0], ciudades[-1], ciudades[1:])
print(len(ciudades), "Málaga" in ciudades)

ciudades.append("Cádiz")          # añade al final
ciudades.insert(1, "Córdoba")     # inserta en una posición
ciudades.extend(["Jaén", "Huelva"])  # añade varios
print(ciudades)

ciudades.remove("Sevilla")        # quita por valor
ultima = ciudades.pop()           # quita y devuelve el último
print(ultima, ciudades)
````

**Ordenar.** `lista.sort()` ordena la lista **en el sitio** (la modifica y devuelve `None`). `sorted(lista)` devuelve una **lista nueva** ordenada y deja la original intacta.

````python
numeros = [5, 2, 9, 1, 7]

ordenada = sorted(numeros)
print(numeros, ordenada)

numeros.sort(reverse=True)
print(numeros)

print(sum(numeros), min(numeros), max(numeros))
````

**Cuidado: asignar no copia.** `b = a` no crea una lista nueva: `a` y `b` son dos nombres para **la misma lista**. Para tener una copia independiente, usa `.copy()`.

````python
a = [1, 2, 3]
b = a            # mismo objeto
c = a.copy()     # copia independiente

b.append(4)
print("a:", a)   # ¡a también ha cambiado!
print("c:", c)
print(a is b, a is c)
````

## 2. Tuplas

Una tupla (`tuple`) es como una lista, pero **inmutable**: una vez creada no se puede modificar. Se usa para agrupar datos que van juntos, como unas coordenadas o un registro. Se escribe entre paréntesis.

Su gran ventaja es el **desempaquetado**: asignar cada elemento a una variable en una sola línea.

````python
punto = (36.7178, -4.4254)
latitud, longitud = punto           # desempaquetado
print(latitud, longitud)

# punto[0] = 0   # descomenta: TypeError, las tuplas no se pueden modificar

a, b = 1, 2
a, b = b, a                         # intercambio sin variable auxiliar
print(a, b)

primero, *resto = (10, 20, 30, 40)  # el * recoge "lo demás"
print(primero, resto)
````

Cuando una función devuelve varios valores, en realidad devuelve una tupla:

````python
def rango(numeros):
    return min(numeros), max(numeros)


resultado = rango([4, 9, 1])
print(resultado, type(resultado))
minimo, maximo = rango([4, 9, 1])
print(minimo, maximo)
````

## 4. Conjuntos

Un conjunto (`set`) es una colección **sin orden** y **sin elementos repetidos**. Es ideal para quitar duplicados y para las operaciones de la teoría de conjuntos.

````python
barrios = ["Centro", "Este", "Centro", "Teatinos", "Este", "Centro"]
unicos = set(barrios)
print(unicos, len(unicos))

a = {1, 2, 3, 4}
b = {3, 4, 5}
print("Unión:", a | b)
print("Intersección:", a & b)
print("Diferencia:", a - b)
print("Diferencia simétrica:", a ^ b)

unicos.add("Huelin")
print("Huelin" in unicos)
````

**¿Por qué importa?** Con `in` puedes preguntar si un elemento está tanto en una lista como en un conjunto, y el resultado es el mismo. Pero ¿tardan lo mismo? Vamos a medirlo: guardamos el mismo millón de números en una lista y en un conjunto, buscamos el último en los dos y cronometramos cuánto tarda cada uno.

````python
import time

def medir(coleccion, elemento, repeticiones=100):
    """Tiempo medio, en microsegundos, de comprobar si `elemento` está en `coleccion`."""
    inicio = time.perf_counter()
    for _ in range(repeticiones):
        elemento in coleccion
    return (time.perf_counter() - inicio) / repeticiones * 1_000_000

numeros_lista = list(range(1_000_000))
numeros_set = set(numeros_lista)
buscado = 999_999

print(buscado in numeros_lista, buscado in numeros_set)   # mismo resultado...

t_lista = medir(numeros_lista, buscado)
t_set = medir(numeros_set, buscado)
print(f"Lista:    {t_lista:10.2f} µs")                      # ...pero no el mismo tiempo
print(f"Conjunto: {t_set:10.2f} µs")
print(f"El conjunto es unas {t_lista / t_set:,.0f} veces más rápido")
````

Ahora repetimos la búsqueda con colecciones cada vez más grandes. Fíjate en cómo cambia el tiempo de cada una:

````python
tamanos = [1_000, 10_000, 100_000, 1_000_000]
resultados = []
for n in tamanos:
    lista = list(range(n))
    conjunto = set(lista)
    resultados.append((n, medir(lista, n - 1), medir(conjunto, n - 1)))

maximo = max(t_lista for _, t_lista, _ in resultados)
for n, t_lista, t_set in resultados:
    print(f"{n:>9,} elementos")
    print(f"   lista    {t_lista:9.2f} µs  " + "█" * round(t_lista / maximo * 40))
    print(f"   conjunto {t_set:9.2f} µs  " + "█" * round(t_set / maximo * 40))
````

## 5. Comprensiones y ordenación

Una **comprensión** crea una lista, un diccionario o un conjunto a partir de otra colección en una sola línea. Se lee casi como una frase: *"x al cuadrado para cada x en números, si x es par"*.

```python
[expresion for elemento in coleccion if condicion]
```

````python
numeros = [1, 2, 3, 4, 5, 6]

# Con un bucle
cuadrados = []
for x in numeros:
    cuadrados.append(x ** 2)

# Con una comprensión: lo mismo en una línea
cuadrados = [x ** 2 for x in numeros]
pares = [x for x in numeros if x % 2 == 0]
print(cuadrados, pares)

ciudades = ["Málaga", "Sevilla", "Granada", "Málaga"]
longitudes = {c: len(c) for c in ciudades}       # diccionario
iniciales = {c[0] for c in ciudades}             # conjunto
print(longitudes, iniciales)
````

### Funciones lambda

Ya sabes definir funciones con `def`. Una **función lambda** es otra forma de escribir una función: más corta, **sin nombre** y en **una sola línea**. Vamos a escribir la misma función de las dos maneras.

```python
lambda parámetros: expresión
```

La lambda no lleva `return`: devuelve automáticamente el resultado de la expresión.

````python
# Función normal
def doble(x):
    return x * 2

# La misma función como lambda
doble_lambda = lambda x: x * 2

print(doble(5), doble_lambda(5))

# Con dos parámetros
def sumar(a, b):
    return a + b

sumar_lambda = lambda a, b: a + b
print(sumar(3, 4), sumar_lambda(3, 4))

# Con una condición (una sola expresión, usando "valor_si if condición else valor_no")
def paridad(n):
    if n % 2 == 0:
        return "par"
    return "impar"

paridad_lambda = lambda n: "par" if n % 2 == 0 else "impar"
print(paridad(7), paridad_lambda(7))
````

**¿Qué ventaja tiene?** Guardar una lambda en una variable, como en la celda anterior, solo sirve para compararla con `def`; para eso es mejor usar `def`. La lambda es útil cuando una función **necesita recibir otra función** como argumento y esa función es tan pequeña que no merece un nombre ni una definición aparte: la escribes **justo donde la usas**.

Su límite es que solo admite **una expresión**: nada de varias líneas, bucles ni `if` con bloques. Si la lógica es más complicada, o vas a reutilizar la función, usa `def`.

El caso más habitual de "función que recibe otra función" es ordenar, que vemos a continuación.

### Ordenar con criterio: el parámetro `key`

**Paso 1. Ordenar lo sencillo.** `sorted()` ordena números de menor a mayor y textos alfabéticamente, sin que le digamos nada. `min()` y `max()` también.

````python
print(sorted([30, 16, 20]))
print(sorted(["Soho", "Muelle Uno", "Alameda Principal"]))
print(max([30, 16, 20]))
````

**Paso 2. El problema.** ¿Y si cada elemento es un diccionario? Python no sabe si una estación es "mayor" que otra: ¿por nombre?, ¿por capacidad? Si lo intentas, da `TypeError`.

````python
estaciones = [
    {"nombre": "Muelle Uno", "capacidad": 20},
    {"nombre": "Alameda Principal", "capacidad": 30},
    {"nombre": "Soho", "capacidad": 16},
]

# sorted(estaciones)   # descomenta: TypeError, no sabe comparar diccionarios
````

**Paso 3. La solución: `key`.** `sorted()`, `min()` y `max()` aceptan un parámetro `key` al que le pasas **una función**. Python la aplica a **cada elemento** y ordena según lo que devuelve, pero te devuelve los **elementos originales**. Primero lo hacemos con una función normal:

````python
def capacidad(estacion):
    return estacion["capacidad"]

print(capacidad(estaciones[0]))       # lo que "ve" sorted() de cada estación: 20

por_capacidad = sorted(estaciones, key=capacidad)   # sin paréntesis: se la damos, no la llamamos
print([e["nombre"] for e in por_capacidad])
````

**Paso 4. Lo mismo con una lambda.** La función `capacidad` solo la queremos para ordenar, así que podemos escribirla como lambda directamente dentro de `key`. Con `reverse=True` el orden se invierte.

````python
por_capacidad = sorted(estaciones, key=lambda e: e["capacidad"], reverse=True)
print([e["nombre"] for e in por_capacidad])

mas_pequena = min(estaciones, key=lambda e: e["capacidad"])
print(mas_pequena["nombre"])          # devuelve la estación entera, no el 16

# key también acepta funciones que ya existen
print(sorted(["pera", "Kiwi", "manzana"], key=str.lower))  # sin distinguir mayúsculas
print(sorted(["pera", "Kiwi", "manzana"], key=len))        # por longitud
````

### CSV

Un CSV es una tabla en texto plano: una fila por línea y columnas separadas por comas. El módulo `csv` lo lee correctamente, incluso con comas dentro de comillas. `csv.DictReader` devuelve cada fila como un **diccionario** con los nombres de las columnas.

**Ojo:** todo lo que se lee de un CSV es **texto** (`str`), aunque parezca un número.

````python
import csv

with open("datos/viajes.csv", encoding="utf-8", newline="") as f:
    lector = csv.DictReader(f)
    print(lector.fieldnames)
    for i, fila in enumerate(lector):
        print(fila)
        if i == 2:
            break

print(type(fila["duracion_min"]))   # str, no float
````

Para escribir un CSV, `csv.DictWriter` hace el camino inverso:

````python
filas = [
    {"ciudad": "Málaga", "habitantes": 592000},
    {"ciudad": "Sevilla", "habitantes": 684000},
]

with open("salida/ciudades.csv", "w", encoding="utf-8", newline="") as f:
    escritor = csv.DictWriter(f, fieldnames=["ciudad", "habitantes"])
    escritor.writeheader()
    escritor.writerows(filas)

print(Path("salida/ciudades.csv").read_text(encoding="utf-8"))
````

### JSON

JSON es el formato estándar para intercambiar datos en la web y entre programas. Se parece mucho a los diccionarios y listas de Python, y el módulo `json` los convierte en un sentido y en otro.

| Python | JSON |
|---|---|
| `dict` | objeto `{...}` |
| `list` | array `[...]` |
| `str`, `int`, `float` | cadena, número |
| `True` / `False` / `None` | `true` / `false` / `null` |

````python
import json

with open("datos/estaciones.json", encoding="utf-8") as f:
    contenido = json.load(f)              # JSON → Python

print(contenido.keys())
print(len(contenido["estaciones"]), "estaciones")
print(contenido["estaciones"][0])

resumen = {"ciudad": "Málaga", "estaciones": 12, "abierto": True}
with open("salida/prueba.json", "w", encoding="utf-8") as f:
    json.dump(resumen, f, ensure_ascii=False, indent=2)   # Python → JSON

print(Path("salida/prueba.json").read_text(encoding="utf-8"))
````

## 7. Errores y excepciones

Cuando algo va mal, Python lanza una **excepción**. Si nadie la captura, el programa se detiene. Ya has visto varias:

| Excepción | Cuándo ocurre |
|---|---|
| `ValueError` | Un valor no tiene sentido: `int("N/D")` |
| `TypeError` | Operación con un tipo incorrecto: `"3" + 4` |
| `KeyError` | La clave no existe en un diccionario |
| `IndexError` | La posición no existe en una lista |
| `ZeroDivisionError` | División entre cero |
| `FileNotFoundError` | El fichero no existe |

Con `try`/`except` puedes **capturar** el error y decidir qué hacer, en lugar de que el programa se detenga. Es imprescindible al trabajar con datos reales, que casi siempre vienen "sucios".

````python
for texto in ["12.5", "N/D", "", "-3"]:
    try:
        valor = float(texto)
    except ValueError:
        print(f"{texto!r}: no es un número")
    else:                       # solo si no hubo excepción
        print(f"{texto!r}: {valor}")
    finally:                    # siempre, haya error o no
        pass
````

**Captura solo los errores que esperas.** Un `except:` sin tipo lo atrapa todo, incluidos tus propios fallos de programación, y los esconde. Indica siempre el tipo de excepción.

También puedes **lanzar** tus propias excepciones con `raise`, por ejemplo para validar datos, e incluso definir tipos nuevos:

````python
class DatoInvalido(Exception):
    """Error para datos que no cumplen las reglas del negocio."""


def validar_duracion(minutos):
    if minutos <= 0:
        raise DatoInvalido(f"La duración debe ser positiva y es {minutos}")
    return minutos


try:
    validar_duracion(-4)
except DatoInvalido as error:
    print("Viaje descartado:", error)
````

## 8. Biblioteca estándar y módulos

### `collections`: contar y agrupar

````python
from collections import Counter, defaultdict, namedtuple

tipos = ["abonado", "ocasional", "abonado", "abonado", "ocasional"]
conteo = Counter(tipos)
print(conteo, conteo.most_common(1))

# defaultdict crea un valor por defecto la primera vez que usas una clave
por_barrio = defaultdict(list)
for estacion, barrio in [("Alameda", "Centro"), ("El Palo", "Este"), ("Soho", "Centro")]:
    por_barrio[barrio].append(estacion)
print(dict(por_barrio))

# namedtuple: tupla con nombres de campo
Viaje = namedtuple("Viaje", ["origen", "destino", "minutos"])
v = Viaje("Alameda", "Soho", 12.5)
print(v.origen, v.minutos)
````

### `datetime`: fechas y horas

````python
from datetime import date, datetime, timedelta

fecha = datetime.strptime("2026-09-18 08:45", "%Y-%m-%d %H:%M")   # texto → fecha
print(fecha, fecha.year, fecha.hour)
print(fecha.weekday())          # 0 = lunes ... 6 = domingo
print(fecha + timedelta(days=7, minutes=30))
print(fecha.strftime("%d/%m/%Y"))   # fecha → texto

dias = (date(2026, 12, 25) - date(2026, 10, 1)).days
print(dias, "días hasta Navidad")
````
