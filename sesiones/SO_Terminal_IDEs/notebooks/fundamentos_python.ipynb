## 1. ¿Qué es Python?

Python es un lenguaje de programación:

- **Interpretado:** no hay que compilar. El intérprete lee y ejecuta el código línea a línea.
- **De tipado dinámico:** las variables no se declaran con un tipo. El tipo lo tiene el valor, no la variable.
- **Legible:** los bloques se marcan con la **sangría** (4 espacios), no con llaves.

| | Java | Python |
|---|---|---|
| Ejecución | Se compila (`javac`) y luego se ejecuta (`java`) | Se ejecuta directamente (`python archivo.py`) |
| Tipos | `int x = 5;` | `x = 5` |
| Bloques | `{ ... }` | Sangría y `:` |
| Fin de instrucción | `;` | Salto de línea |
| Hola mundo | Una clase con un método `main` | `print("Hola")` |

Hay tres formas de ejecutar Python: en el **intérprete interactivo** (REPL, escribiendo `python3` en la terminal), como **script** (`python3 archivo.py`) o en un **notebook** como este, que mezcla texto, código y resultados.

## 3. Cadenas de texto

Una cadena (`str`) es una secuencia de caracteres. Cada carácter tiene una **posición** (índice) que empieza en `0`. Los índices negativos cuentan desde el final.

```
 P   y   t   h   o   n
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

````python
lenguaje = "Python"

print(lenguaje[0], lenguaje[-1])   # primero y último
print(lenguaje[0:3])               # del 0 al 2 (el 3 no se incluye)
print(lenguaje[2:])                # desde el 2 hasta el final
print(lenguaje[::-1])              # al revés
print(len(lenguaje))               # longitud
print("th" in lenguaje)            # ¿contiene "th"?
````

**Métodos de las cadenas.** Las cadenas son **inmutables**: los métodos no cambian la cadena original, devuelven una nueva.

````python
texto = "  Big Data en Málaga  "

print(texto.strip())                  # quita espacios al principio y al final
print(texto.upper(), texto.lower())
print(texto.replace("Málaga", "UMA"))
print(texto.split())                  # trocea por espacios → lista
print("-".join(["2026", "10", "05"])) # une una lista con un separador
print(texto.strip().startswith("Big"))
````

**f-strings.** La forma moderna de mezclar texto y variables: se pone una `f` delante de las comillas y las expresiones entre `{}`. Después de `:` se indica el formato.

````python
nombre = "Ada"
saldo = 1234.5678

print(f"Hola, {nombre}. Tu saldo es {saldo:.2f} €")
print(f"Con separador de miles: {1234567:,}")
print(f"Porcentaje: {0.256:.1%}")
print(f"[{nombre:>10}]")   # alineado a la derecha en 10 caracteres
print(f"{2 + 3 = }")       # muestra la expresión y su resultado
````

## 4. Control de flujo

### Condicionales: `if`, `elif`, `else`

Fíjate en los `:` al final de cada condición y en la sangría del bloque.

````python
nota = 7.5

if nota >= 9:
    calificacion = "Sobresaliente"
elif nota >= 7:
    calificacion = "Notable"
elif nota >= 5:
    calificacion = "Aprobado"
else:
    calificacion = "Suspenso"

print(calificacion)
````

### `match`

Compara un valor con varios patrones. Es útil cuando hay muchas opciones fijas. `_` recoge cualquier otro caso.

````python
comando = "stop"

match comando:
    case "start":
        print("Arrancando")
    case "stop" | "exit":
        print("Parando")
    case _:
        print(f"Comando desconocido: {comando}")
````

### Bucles: `while` y `for`

- `while` repite **mientras** se cumpla una condición.
- `for` recorre **cada elemento** de una secuencia (lista, cadena, rango...).

````python
cuenta = 3
while cuenta > 0:
    print(cuenta)
    cuenta -= 1
print("¡Despegue!")
````

````python
for i in range(5):            # 0, 1, 2, 3, 4
    print(i, end=" ")
print()

for i in range(2, 11, 2):     # del 2 al 10 de 2 en 2
    print(i, end=" ")
print()

for letra in "Python":
    print(letra, end="-")
````

`enumerate()` da a la vez la posición y el elemento. `zip()` recorre varias listas en paralelo.

````python
ciudades = ["Málaga", "Sevilla", "Granada"]
habitantes = [592_000, 684_000, 228_000]

for i, ciudad in enumerate(ciudades, start=1):
    print(i, ciudad)

for ciudad, n in zip(ciudades, habitantes):
    print(f"{ciudad}: {n:,} habitantes")
````

`break` sale del bucle, `continue` salta a la siguiente vuelta. El `else` de un bucle se ejecuta solo si el bucle terminó **sin** `break`.

````python
numeros = [3, 8, 11, 14, 21]

for n in numeros:
    if n % 2 == 0:
        continue          # salta los pares
    if n > 20:
        print("Encontrado un impar mayor que 20:", n)
        break
else:
    print("No hay ningún impar mayor que 20")
````

## 5. Funciones

Una función agrupa código con un nombre para reutilizarlo. Se define con `def`, recibe **parámetros** y devuelve un resultado con `return`. El texto entre triples comillas es la **docstring**: explica qué hace la función.

````python
def saludar(nombre, saludo="Hola"):
    """Devuelve un saludo personalizado."""
    return f"{saludo}, {nombre}"


print(saludar("Ada"))                      # usa el valor por defecto
print(saludar("Alan", "Buenas tardes"))    # por posición
print(saludar(saludo="Hey", nombre="Grace"))  # por nombre
help(saludar)
````

Una función puede devolver **varios valores** (en realidad, una tupla) y recibir un número variable de argumentos con `*args` (posicionales) y `**kwargs` (con nombre).

````python
def min_max(numeros):
    """Devuelve el mínimo y el máximo de una lista."""
    return min(numeros), max(numeros)


minimo, maximo = min_max([4, 9, 1, 7])
print(minimo, maximo)


def sumar_todo(*args):
    return sum(args)


def describir(**kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave} = {valor}")


print(sumar_todo(1, 2, 3, 4))
describir(modulo="Laboratorio", horas=3)
````

**Ámbito.** Las variables creadas dentro de una función son **locales**: no existen fuera de ella.

````python
def doble(x):
    resultado = x * 2
    return resultado


print(doble(5))
# print(resultado)  # descomenta: NameError, "resultado" solo existe dentro de doble()
````

**Recursividad.** Una función puede llamarse a sí misma. Necesita un **caso base** que detenga las llamadas.

````python
def factorial(n):
    """n! = n · (n-1) · ... · 1"""
    if n <= 1:          # caso base
        return 1
    return n * factorial(n - 1)


print(factorial(5))
````

## 7. Depurar con VS Code

La función `media()` tiene un error: `media([2, 4, 6])` debería dar `4.0`, pero da otra cosa.

Para encontrarlo con el **depurador**:
1. Haz clic a la izquierda del número de la línea del `for` para poner un *breakpoint* (aparece un punto rojo). Si no ves los números de línea, pulsa `L` con la celda seleccionada.
2. Despliega la flecha junto al botón de ejecutar de la celda y elige **Depurar celda** (*Debug Cell*).
3. La ejecución se para en el *breakpoint*. Avanza línea a línea con `F10` y observa en el panel **Variables** qué valores toman `i` y `total`.
4. Cuando encuentres el error, corrígelo y ejecuta la comprobación.

````python
def media(numeros):
    total = 0
    for i in range(1, len(numeros)):
        total += numeros[i]
    return total / len(numeros)


print(media([2, 4, 6]))
````

````python
# Comprobación
assert media([2, 4, 6]) == 4.0
assert media([10, 20]) == 15.0
print("Correcto")
````
